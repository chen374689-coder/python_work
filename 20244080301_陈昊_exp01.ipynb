## 实验 2：函数、条件、循环与递归

`score_level` 对超出范围的分数抛出 `ValueError`。阶乘仅接受非负整数。

```python
def score_level(score):
    if not 0 <= score <= 100:
        raise ValueError("分数必须在 0 到 100 之间")
    if score >= 90:
        return "优秀"
    if score >= 80:
        return "良好"
    if score >= 60:
        return "及格"
    return "不及格"


for score in (100, 90, 89, 80, 79, 60, 59, 0):
    print(score, score_level(score))
try:
    score_level(101)
except ValueError as error:
    print("边界检查:", error)

total_for = 0
for number in range(1, 101):
    total_for += number

total_while = 0
number = 1
while number <= 100:
    total_while += number
    number += 1
print("for:", total_for, "while:", total_while)
assert total_for == total_while == 5050


def factorial_recursive(n):
    if not isinstance(n, int) or isinstance(n, bool) or n < 0:
        raise ValueError("n 必须是非负整数")
    if n <= 1:
        return 1
    return n * factorial_recursive(n - 1)


def factorial_iterative(n):
    if not isinstance(n, int) or isinstance(n, bool) or n < 0:
        raise ValueError("n 必须是非负整数")
    result = 1
    for value in range(2, n + 1):
        result *= value
    return result


for n in (0, 1, 5, 10):
    print(n, factorial_recursive(n), factorial_iterative(n))


def summary(values):
    if not values:
        raise ValueError("values 不能为空")
    return min(values), max(values), sum(values) / len(values)


minimum, maximum, mean = summary([4, 7, 10])
print("summary:", minimum, maximum, mean)


def sum_all(*args):
    return sum(args)


print(sum_all(), sum_all(5), sum_all(1, 2, 3, 4))
```

## 实验 4：文件读写、异常处理与模块

解析器检查字段数量、姓名和分数范围；错误记录被记录下来，后续行仍会继续处理。

```python
import os
import pickle

data_dir = os.path.join(os.getcwd(), "exp01_data")
os.makedirs(data_dir, exist_ok=True)
students_path = os.path.join(data_dir, "students.txt")
source_text = (
    "name,score\nAlice,85\nBob,58\nCarol,92\nDavid,76\n"
    "Eva,60\nFrank,89\nGrace,abc\nHenry,105\nIncomplete\n"
)

with open(students_path, "w", encoding="utf-8") as file:
    file.write(source_text)


def parse_student(line):
    fields = line.strip().split(",")
    if len(fields) != 2:
        raise ValueError("字段数量必须为 2")
    name = fields[0].strip()
    if not name:
        raise ValueError("姓名不能为空")
    try:
        score = int(fields[1].strip())
    except ValueError as error:
        raise ValueError("成绩必须是整数") from error
    if not 0 <= score <= 100:
        raise ValueError("成绩必须在 0 到 100 之间")
    return {"name": name, "score": score}


valid_records = []
invalid_records = []
with open(students_path, encoding="utf-8") as file:
    next(file, None)
    for line_number, line in enumerate(file, start=2):
        try:
            valid_records.append(parse_student(line))
        except ValueError as error:
            invalid_records.append({"line": line_number, "reason": str(error)})

print("有效记录:", valid_records)
print("无效记录:", invalid_records)
average_valid = sum(item["score"] for item in valid_records) / len(valid_records)
pass_rate_valid = sum(item["score"] >= 60 for item in valid_records) / len(valid_records)
report_path = os.path.join(data_dir, "report.txt")
report = (
    f"有效记录数: {len(valid_records)}\n"
    f"无效记录数: {len(invalid_records)}\n"
    f"平均分: {average_valid:.2f}\n"
    f"及格率: {pass_rate_valid:.0%}\n"
)

with open(report_path, "w", encoding="utf-8") as file:
    file.write("旧内容（会被 w 模式覆盖）\n")
with open(report_path, "w", encoding="utf-8") as file:
    file.write(report)
with open(report_path, "a", encoding="utf-8") as file:
    file.write("追加说明：w 覆盖原内容，a 保留原内容并追加。\n")
with open(report_path, encoding="utf-8") as file:
    print(file.read())

pickle_path = os.path.join(data_dir, "students.pkl")
with open(pickle_path, "wb") as file:
    pickle.dump(valid_records, file)
with open(pickle_path, "rb") as file:
    restored_records = pickle.load(file)
assert restored_records == valid_records
print("pickle 恢复数据一致:", restored_records == valid_records)

module_source = '''def parse_student(line):
    fields = line.strip().split(",")
    if len(fields) != 2:
        raise ValueError("字段数量必须为 2")
    name = fields[0].strip()
    if not name:
        raise ValueError("姓名不能为空")
    try:
        score = int(fields[1].strip())
    except ValueError as error:
        raise ValueError("成绩必须是整数") from error
    if not 0 <= score <= 100:
        raise ValueError("成绩必须在 0 到 100 之间")
    return {"name": name, "score": score}
'''
module_path = os.path.join(os.getcwd(), "student_utils.py")
with open(module_path, "w", encoding="utf-8") as file:
    file.write(module_source)

import importlib.util

spec = importlib.util.spec_from_file_location("student_utils", module_path)
student_utils = importlib.util.module_from_spec(spec)
spec.loader.exec_module(student_utils)
print("模块有效输入:", student_utils.parse_student("Ivy,88"))
try:
    student_utils.parse_student("Jack,abc")
except ValueError as error:
    print("模块无效输入:", error)

missing_path = os.path.join(data_dir, "does_not_exist.txt")
try:
    with open(missing_path, encoding="utf-8") as file:
        file.read()
except FileNotFoundError as error:
    print("文件不存在，已捕获异常:", error.filename)
```

## 实验 5：类、继承与对象复制

```python
import copy


class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"


students = [Student(item["name"], item["score"]) for item in valid_records]
for student in students:
    print(student, "及格:", student.is_passed())


class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    def __str__(self):
        return f"GraduateStudent(name={self.name}, score={self.score}, topic={self.research_topic})"


graduate = GraduateStudent("Ivy", 95, "数据分析")
print(graduate, "及格:", graduate.is_passed())


def fresh_object():
    return {"name": "Alice", "scores": [85, 90]}


direct = fresh_object()
direct["scores"].append(100)
print("直接赋值后:", direct)

shallow_original = fresh_object()
shallow = copy.copy(shallow_original)
shallow["scores"].append(100)
print("浅复制原对象:", shallow_original, "浅复制对象:", shallow)

deep_original = fresh_object()
deep = copy.deepcopy(deep_original)
deep["scores"].append(100)
print("深复制原对象:", deep_original, "深复制对象:", deep)
```
