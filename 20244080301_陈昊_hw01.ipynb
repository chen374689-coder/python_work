## 二、欧几里得算法

每一步用较小问题 `gcd(n, m % n)` 替代 `gcd(m, n)`，余数变成 0 时当前除数就是最大公约数。用 `abs` 统一负数输入的符号，因而结果总是非负。

```python
def gcd(m, n):
    m, n = abs(m), abs(n)
    while n != 0:
        m, n = n, m % n
    return m


for pair in [(2018, 2019), (1200, 300), (5040, 60), (-18, 12), (18, -12)]:
    print(pair, "->", gcd(*pair))
```

负数输入时，先取绝对值再求 GCD，符合“最大正公约数”的常见定义。`gcd(0, 0)` 按此实现返回 0；通常最大公约数的定义要求至少一个输入非零。

## 三、近似欧拉数 `e`

极限近似使用 `euler_limit(n) = (1 + 1/n) ** n`，因此 `n` 必须为正整数。级数函数按题目约定返回前 `n` 项：`euler_infinite_sum(0) == 0`。

```python
import math


def euler_limit(n):
    if n <= 0:
        raise ValueError("n 必须是正整数")
    return (1 + 1 / n) ** n


def euler_infinite_sum(n):
    if n < 0:
        raise ValueError("n 必须是非负整数")
    total = 0.0
    term = 1.0
    for k in range(n):
        if k > 0:
            term /= k
        total += term
    return float(total)


def euler_approx(epsilon):
    if epsilon <= 0:
        raise ValueError("epsilon 必须大于 0")
    total = 0.0
    term = 1.0
    k = 0
    while abs(math.exp(1) - total) > epsilon:
        if k > 0:
            term /= k
        total += term
        k += 1
    return float(total)


def print_euler_sum_table(n):
    if n <= 0:
        raise ValueError("n 必须是正整数")
    for k in range(1, n + 1):
        print(f"n={k}: {euler_infinite_sum(k)}")


print("limit n=10:", euler_limit(10))
print("first 4 terms:", euler_infinite_sum(4))
print("first term:", euler_infinite_sum(1))
print("zero terms:", euler_infinite_sum(0))
print("epsilon=1e-6:", euler_approx(1e-6))
print_euler_sum_table(6)
```

## 四、判断整数幂

逐次用除法和取模去掉底数因子；若最终恰好剩下 1，则原数是该底数的非负整数次幂。

```python
def is_power_of_2(n):
    if n <= 0:
        return False
    while n % 2 == 0:
        n //= 2
    return n == 1


def is_power(b, n):
    if b <= 0 or n <= 0:
        return False
    if b == 1:
        return n == 1
    while n % b == 0:
        n //= b
    return n == 1


print([(n, is_power_of_2(n)) for n in (1, 2, 8, 12, 64)])
print([(b, n, is_power(b, n)) for b, n in ((2, 64), (3, 81), (5, 20), (1, 1), (1, 7))])
```
