# 字符串的乐趣

在这个问题中，您将实现一些用于处理字符串的简单函数。您不需要为这个问题中的任何函数执行错误检查。

1. 回文是指一个词或短语，它从前往后读和从后往前读都一样。例如，level、kayak 和 pop 都是英文回文；如果忽略空格和标点符号，rats live on no evil star 和 Was it a car or a cat I saw? 也是回文短语。编写一个名为 is_palindrome 的函数，它接受一个字符串作为唯一参数，并返回布尔值：如果输入是回文，返回 True，否则返回 False。您可以假设输入仅包含英文字母（不区分大小写）、数字 0 到 9 和空格。判断回文时应忽略空格和大小写，因此 tacocat 和 T A C O cat 都应被认为是回文。

2. 如果一个词的字母从左到右按字母顺序排列，就称它是“字母顺序的”；允许字母重复。例如，adder 和 beet 是字母顺序的，dog 和 cat 则不是。编写一个名为 is_abecedarian 的函数，它只接受一个字符串参数。如果参数符合上述要求，返回 True，否则返回 False。您可以假设输入仅由字母和空格组成。判断时应忽略空格，因此 abcd efgh xyz 被认为是字母顺序的。

3. 编写一个名为 remove_vowels 的函数，它接受一个字符串作为唯一参数，并返回去掉所有元音后的字符串。本题的元音是字母 a、e、i、o、u。例如，remove_vowels('cat') 应返回 'ct'，remove_vowels('audio') 应返回 'd'。函数也应正确处理全部由元音组成的字符串。


```python
# 第 1 题：is_palindrome —— 忽略空格和大小写判断回文
def is_palindrome(s):
    cleaned = "".join(ch.lower() for ch in s if ch != " ")
    return cleaned == cleaned[::-1]

print(is_palindrome("level"))                      # True
print(is_palindrome("T A C O cat"))                # True
print(is_palindrome("rats live on no evil star"))  # True
print(is_palindrome("python"))                     # False
```

```python
# 第 2 题：is_abecedarian —— 忽略空格，判断字母是否按字母顺序排列
def is_abecedarian(s):
    letters = [ch.lower() for ch in s if ch != " "]
    return all(letters[i] <= letters[i + 1] for i in range(len(letters) - 1))

print(is_abecedarian("adder"))          # True
print(is_abecedarian("beet"))           # True
print(is_abecedarian("abcd efgh xyz"))  # True（忽略空格）
print(is_abecedarian("dog"))            # False
```

```python
# 第 3 题：remove_vowels —— 去掉所有元音 a、e、i、o、u
def remove_vowels(s):
    vowels = "aeiou"
    return "".join(ch for ch in s if ch.lower() not in vowels)

print(remove_vowels("cat"))    # ct
print(remove_vowels("audio"))  # d
print(repr(remove_vowels("aeiou")))  # '' 空字符串
```

# 列表的乐趣

在这个问题中，您将实现一些简单的列表操作。

1. 编写一个名为 list_reverse 的函数，它接受一个列表作为参数，并返回反转后的列表。例如，输入 [1, 2, 3]，应返回 [3, 2, 1]。如果输入不是列表，函数应引发适当的错误。

2. 编写一个名为 is_sorted 的函数，它接受一个序列 seq 作为唯一参数。如果序列按升序排列，返回 True，否则返回 False。您可以假设 seq 是 Python 序列，且所有元素都支持比较操作，因此无需检查输入类型。函数只应遍历序列一次；需要多次遍历的低效解法不能获得满分。这个问题展示了 Python 动态类型的特点：只要 seq 支持索引，且元素支持比较，就无需预先知道序列的具体类型。


```python
# 第 1 题：list_reverse —— 返回反转后的新列表；输入不是列表则引发 TypeError
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("list_reverse 的参数必须是列表")
    return lst[::-1]

print(list_reverse([1, 2, 3]))   # [3, 2, 1]
print(list_reverse(["a", "b"]))  # ['b', 'a']
try:
    list_reverse("not a list")
except TypeError as e:
    print("TypeError:", e)
```

```python
# 第 2 题：is_sorted —— 只遍历一次判断升序
def is_sorted(seq):
    for i in range(len(seq) - 1):
        if seq[i + 1] < seq[i]:
            return False
    return True

print(is_sorted([1, 2, 2, 5]))  # True（允许相等）
print(is_sorted((1, 3, 2)))     # False（元组同样适用）
print(is_sorted("abcd"))        # True
```

# 作为向量的元组

在这个问题中，我们使用元组来表示向量。这里先用一个长度为 d 的元组表示 d 维向量。

1. 实现一个名为 vec_scalar_mult 的函数。它接受两个参数：一个由整数或浮点数组成的元组 t，以及一个整数或浮点数 s。函数返回与 t 长度相同的元组，每个元素等于 t 的对应元素乘以 s。函数应检查输入类型；类型不合适时，引发带有适当错误消息的 TypeError。整数标量和含有整数元素的元组都是有效输入。

2. 实现一个名为 vec_inner_product 的函数。它接受两个由整数或浮点数组成的元组，返回这两个向量内积对应的浮点数。向量 x, y 的内积是 Σ x_j·y_j。若任一输入不是元组，应引发 TypeError；若两个元组的长度不一致，应引发 ValueError。

3. 将上述表示方法扩展到矩阵。用一个元组的元组表示 m×n 矩阵 my_mx：外层元组长度为 m，每个内层元组长度为 n，表示矩阵的一行。编写一个名为 check_valid_mx 的函数，它接受一个参数；若该参数是有效矩阵则返回 True，否则返回 False。有效矩阵满足：

   - 外层元组的每个元素本身都是元组；
   - 所有内层元组长度相同；
   - 每个内层元组的每个元素都是整数或浮点数。

4. 编写一个名为 mx_vec_mult 的函数，它接受一个矩阵（元组的元组）和一个向量（元组），返回向量与矩阵相乘得到的数值元组。函数 check_valid_mx 应检查参数是否合理；对于不合理的输入，应引发适当的错误。


```python
# 第 1 题：vec_scalar_mult —— 元组乘以标量，返回新元组
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("第一个参数 t 必须是元组")
    for x in t:
        if not isinstance(x, (int, float)) or isinstance(x, bool):
            raise TypeError("元组 t 的每个元素必须是整数或浮点数")
    if not isinstance(s, (int, float)) or isinstance(s, bool):
        raise TypeError("第二个参数 s 必须是整数或浮点数")
    return tuple(x * s for x in t)

print(vec_scalar_mult((1, 2, 3), 2.5))  # (2.5, 5.0, 7.5)
print(vec_scalar_mult((1, 2), 3))       # (3, 6)：整数标量同样有效
```

```python
# 第 2 题：vec_inner_product —— 两个元组的内积（返回浮点数）
def vec_inner_product(t1, t2):
    if not isinstance(t1, tuple) or not isinstance(t2, tuple):
        raise TypeError("两个参数都必须是元组")
    if len(t1) != len(t2):
        raise ValueError("两个元组的长度必须一致")
    return float(sum(x * y for x, y in zip(t1, t2)))

print(vec_inner_product((1, 2), (3, 4)))   # 11.0
try:
    vec_inner_product((1, 2), (1, 2, 3))
except ValueError as e:
    print("ValueError:", e)
```

```python
# 第 3 题：check_valid_mx —— 判断“元组的元组”是否为有效矩阵
def check_valid_mx(m):
    if not isinstance(m, tuple) or len(m) == 0:
        return False
    for row in m:
        if not isinstance(row, tuple) or len(row) == 0:
            return False
        for x in row:
            if not isinstance(x, (int, float)) or isinstance(x, bool):
                return False
    row_len = len(m[0])
    return all(len(row) == row_len for row in m)

print(check_valid_mx(((1, 2), (3, 4))))    # True
print(check_valid_mx(((1, 2), (3,))))      # False：内层长度不同
print(check_valid_mx(((1, "a"), (2, 3))))  # False：含非数值元素
print(check_valid_mx([(1, 2), (3, 4)]))    # False：外层不是元组
```

```python
# 第 4 题：mx_vec_mult —— 矩阵（元组的元组）乘向量（元组）
def mx_vec_mult(mx, vec):
    if not check_valid_mx(mx):
        raise ValueError("第一个参数不是有效矩阵")
    if not isinstance(vec, tuple):
        raise TypeError("第二个参数必须是元组")
    if len(mx[0]) != len(vec):
        raise ValueError("矩阵列数必须与向量长度一致")
    return tuple(
        sum(row[j] * vec[j] for j in range(len(vec)))
        for row in mx
    )

my_mx = ((1, 2), (3, 4))
print(mx_vec_mult(my_mx, (1, 1)))             # (3, 7)
print(mx_vec_mult(((1, 0), (0, 1)), (5, 6)))  # (5, 6)：单位矩阵
```

# 更多向量的乐趣

在这个练习中，我们将再次遇到向量，这次采用面向对象的方法。

1. 定义一个名为 Vector 的类。每个向量都有一个维度（非负整数）和一个元素列表或元组。初始化函数的第一个参数是维度，第二个参数是由整数或浮点数组成的列表或元组，表示向量的元素。如果只提供维度而不提供元素，请选择合理的默认行为。若维度的类型不正确或为负数，应抛出合适的异常；若维度与元素个数不一致，也应抛出异常。

2. 维度和元素应是类属性还是实例属性？说明选择的理由。

3. 实现向量对象的比较运算符，包括相等、小于、小于或等于、大于等。两个 Vector 对象的坐标相同，则视为相等；否则按 Python 元组的字典序比较，即先比较第一个坐标，再比较第二个坐标，依此类推。例如，二维向量 (2, 4) 小于 (2, 5)。比较维度不同的两个向量时，应抛出异常。

4. 实现 Vector.dot 方法，接收另一个 Vector 对象，返回两向量的内积。若参数类型不正确或维度不一致，应抛出合适的异常。

5. 实现 Vector 的乘法运算符。标量左乘或右乘（如 2*v 和 v*2）应返回一个各元素都乘以该标量的新 Vector 对象。当两个 Vector 对象相乘（v*w）时，对应位置的元素逐一相乘，结果仍是一个新 Vector 对象；例如，Vector(2, (2, 3)) * Vector(2, (4, 5)) 的元素为 (8, 15)。若两个向量的维度不一致，应抛出合适的异常。


```python
# 第 1 题：Vector 类的基本定义
# 只给维度、不给元素时，默认为该维度的零向量
class Vector:
    def __init__(self, dim, coords=None):
        if not isinstance(dim, int) or isinstance(dim, bool):
            raise TypeError("维度必须是整数")
        if dim < 0:
            raise ValueError("维度不能为负数")
        if coords is None:
            coords = (0.0,) * dim
        if not isinstance(coords, (list, tuple)):
            raise TypeError("元素必须是列表或元组")
        if len(coords) != dim:
            raise ValueError("元素个数必须与维度一致")
        for x in coords:
            if not isinstance(x, (int, float)) or isinstance(x, bool):
                raise TypeError("每个元素必须是整数或浮点数")
        # 维度和元素都保存为实例属性
        self.dim = dim
        self.coords = tuple(coords)

    def __repr__(self):
        return f"Vector({self.dim}, {self.coords})"

print(Vector(2, (2, 3)))  # Vector(2, (2, 3))
print(Vector(3))          # Vector(3, (0.0, 0.0, 0.0))
```

**第 2 题：维度和元素应是实例属性。**

理由：每个向量的维度和坐标都各不相同，属于各个对象自身的数据。如果把它们写成类属性，所有 Vector 实例就会共享同一个 `coords`，创建第二个向量时会把第一个向量的坐标覆盖掉，`v == w` 之类的判断也会失真。因此 `self.dim` 和 `self.coords` 都在 `__init__` 中赋值为实例属性。类属性只适合存放所有实例共用的常量或默认值。


```python
# 第 3 题：比较运算符（按坐标的字典序比较）
# 在上一格已定义的 Vector 基础上继承扩展，无须重复编写 __init__
class Vector(Vector):
    def _check_dim(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能与另一个 Vector 比较")
        if self.dim != other.dim:
            raise ValueError("维度不同的向量不能比较")

    def __eq__(self, other):
        if not isinstance(other, Vector):
            return NotImplemented
        if self.dim != other.dim:
            raise ValueError("维度不同的向量不能比较")
        return self.coords == other.coords

    def __lt__(self, other):
        self._check_dim(other)
        return self.coords < other.coords

    def __le__(self, other):
        self._check_dim(other)
        return self.coords <= other.coords

    def __gt__(self, other):
        self._check_dim(other)
        return self.coords > other.coords

    def __ge__(self, other):
        self._check_dim(other)
        return self.coords >= other.coords

print(Vector(2, (2, 4)) < Vector(2, (2, 5)))    # True：按坐标字典序
print(Vector(2, (2, 3)) == Vector(2, (2, 3)))   # True：坐标相同
try:
    Vector(2, (1, 2)) < Vector(3, (1, 2, 3))
except ValueError as e:
    print("ValueError:", e)
```

```python
# 第 4 题：Vector.dot —— 两个向量的内积
class Vector(Vector):
    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("dot 的参数必须是 Vector 对象")
        if self.dim != other.dim:
            raise ValueError("维度不一致，无法求内积")
        return float(sum(x * y for x, y in zip(self.coords, other.coords)))

print(Vector(2, (2, 3)).dot(Vector(2, (4, 5))))  # 23.0
try:
    Vector(2, (1, 2)).dot((1, 2))
except TypeError as e:
    print("TypeError:", e)
```

```python
# 第 5 题：乘法运算符 —— 标量左乘/右乘，以及两个向量逐元素相乘
class Vector(Vector):
    def __mul__(self, other):
        if isinstance(other, (int, float)) and not isinstance(other, bool):
            return Vector(self.dim, tuple(x * other for x in self.coords))
        if isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("维度不一致，无法逐元素相乘")
            return Vector(self.dim, tuple(x * y for x, y in zip(self.coords, other.coords)))
        raise TypeError("只能乘以标量或另一个 Vector")

    def __rmul__(self, other):
        # 支持 2 * v（标量在左）
        return self.__mul__(other)

v = Vector(2, (2, 3))
w = Vector(2, (4, 5))
print(2 * v)  # Vector(2, (4, 6))：标量左乘
print(v * 2)  # Vector(2, (4, 6))：标量右乘
print(v * w)  # Vector(2, (8, 15))：逐元素相乘
```
