# Python 正则表达式练习
请编写 Python 函数完成以下任务，并⽤测试⽤例展⽰。
1. 检查字符串是否仅包含英⽂字⺟ a-z、A-Z 和数字 0-9。
2. 匹配⼀个 a 后⾯跟着零个或多个 b 的字符串。
3. 匹配⼀个 a 后⾯跟着⼀个或多个 b 的字符串。
4. 匹配以 a 开头、以 b 结尾，且中间可以有任意字符的字符串。
5. 删除 IP 地址中各段数字的前导零。例如，192.168.001.010 应变为
192.168.1.10；某段全为零时保留⼀个 0。
6. 将字符串中的空格替换为下划线，并将下划线替换为空格。
7. 将 yyyy-mm-dd 格式的⽇期转换为 dd-mm-yyyy 格式。
8. 找出给定字符串中所有以 a 或 e 开头的单词。
9. 将给定字符串中的 Road 缩写为 Rd。
10. 找出字符串中所有⻓度⾄少为 4 个字符的单词。
11. 删除字符串中除字⺟和数字之外的所有字符。
12. 删除字符串中的括号及括号内的内容。

例如，对下⾯的输⼊逐项处理：
example (.com)
w3resource
github (.com)
stackoverflow (.com)

预期依次得到：example、w3resource、github、stackoverflow。

```python
import re

# 1. 检查字符串是否仅包含英文字母 a-z、A-Z 和数字 0-9
def is_alnum_only(s: str) -> bool:
    # 不用 \w 是因为 \w 还包含下划线 _ 和非 ASCII 字符
    return bool(re.fullmatch(r"[a-zA-Z0-9]+", s))
print("1. 仅字母和数字:", [is_alnum_only(x) for x in ["Abc123", "abc_123", "hello world", ""]])


# 2. 匹配一个 a 后面跟着零个或多个 b 的字符串
def match_a_zero_or_more_b(s: str) -> bool:
    return bool(re.fullmatch(r"ab*", s))
print("2. a 后跟 0+ 个 b:", [match_a_zero_or_more_b(x) for x in ["a", "ab", "abbb", "ac", "ba"]])


# 3. 匹配一个 a 后面跟着一个或多个 b 的字符串
def match_a_one_or_more_b(s: str) -> bool:
    return bool(re.fullmatch(r"ab+", s))
print("3. a 后跟 1+ 个 b:", [match_a_one_or_more_b(x) for x in ["a", "ab", "abbb", "ac"]])


# 4. 匹配以 a 开头、以 b 结尾，且中间可以有任意字符的字符串
def match_a_anything_b(s: str) -> bool:
    return bool(re.fullmatch(r"a.*b", s, flags=re.DOTALL))
print("4. a 开头 b 结尾:", [match_a_anything_b(x) for x in ["ab", "a123_xyz b", "a\nb", "ba"]])


# 5. 删除 IP 地址中各段数字的前导零（某段全为零时保留一个 0）
def remove_ip_leading_zeros(ip: str) -> str:
    # 0+ 贪婪匹配前导零，(\d) 至少保留该段最后一位数字
    return re.sub(r"\b0+(\d)", r"\1", ip)
print(
    "5. 去除 IP 前导零:",
    [remove_ip_leading_zeros(x) for x in ["192.168.001.010", "000.00.0.01", "010.000.001.100"]],
)


# 6. 将字符串中的空格替换为下划线，并将下划线替换为空格
def swap_space_and_underscore(s: str) -> str:
    # 必须一次替换完成，避免分两步导致互相覆盖
    return re.sub(r"[ _]", lambda m: "_" if m.group(0) == " " else " ", s)
print("6. 空格与下划线互换:", swap_space_and_underscore("hello_world python_regex test"))


# 7. 将 yyyy-mm-dd 格式的日期转换为 dd-mm-yyyy 格式
def convert_date_format(s: str) -> str:
    return re.sub(r"\b(\d{4})-(\d{2})-(\d{2})\b", r"\3-\2-\1", s)
print("7. 日期格式转换:", convert_date_format("Date: 2026-11-24 and 1999-01-05"))


# 8. 找出给定字符串中所有以 a 或 e 开头的单词
def find_words_starting_with_a_or_e(s: str) -> list[str]:
    return re.findall(r"\b[ae][a-zA-Z]*\b", s, flags=re.IGNORECASE)
print(
    "8. a/e 开头单词:",
    find_words_starting_with_a_or_e("An apple and an elephant eat every apricot."),
)


# 9. 将给定字符串中的 Road 缩写为 Rd
def abbreviate_road(s: str) -> str:
    return re.sub(r"\bRoad\b", "Rd", s)
print("9. Road -> Rd:", abbreviate_road("123 Main Road, Broad Roadway, Park Road."))


# 10. 找出字符串中所有长度至少为 4 个字符的单词
def find_words_len_at_least_4(s: str) -> list[str]:
    return re.findall(r"\b[a-zA-Z]{4,}\b", s)
print(
    "10. 长度>=4 的单词:",
    find_words_len_at_least_4("I saw a quick brown fox jump over the lazy dog."),
)


# 11. 删除字符串中除字母和数字之外的所有字符
def remove_non_alnum(s: str) -> str:
    return re.sub(r"[^a-zA-Z0-9]", "", s)
print("11. 删除非字母数字:", remove_non_alnum("Hello, World! 2026_@#Python-3.12"))

samples_12 = ["example (.com)", "w3resource", "github (.com)", "stackoverflow (.com)"]


# 12. 删除字符串中的括号及括号内的内容
def remove_parentheses(s: str) -> str:
    # 连同括号前的多余空白一起去掉
    return re.sub(r"\s*\([^)]*\)", "", s).strip()
print("12. 删除括号及内容:", [remove_parentheses(x) for x in samples_12])
```

3. 编写 list_climate_table_row_names(s)，只接收⼀个城市⻚⾯ URL。若找
到⽓候数据表格，返回其⾏名组成的列表；若成功读取⻚⾯但找不到，返回
None。列表中的每项都必须是去除⾸尾空⽩的 Python 字符串，不能包含
BeautifulSoup 对象或 HTML 标签。不要把“Climate data for...”标题⾏、⽉份列标题
和数据来源⾏计⼊结果。提⽰：结合 tr 的结构与 th 的位置判断⾏名；如有
scope="row"，可⽤它辅助判断，但不要假定每个⻚⾯都有该属性。

```python
import datetime
import re
import requests
from bs4 import BeautifulSoup, Tag

URL = "https://www.weather.com.cn/weather/101250401.shtml"
HEADERS = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/120.0.0.0 Safari/537.36"
    )
}


def _normalize_space(text: str) -> str:
    """合并连续空白字符并去除首尾空白。"""
    return re.sub(r"\s+", " ", text).strip()


def _fetch_soup(url: str) -> BeautifulSoup:
    """获取网页并用 html.parser 解析；网络或 HTTP 错误直接抛出异常。"""
    resp = requests.get(url, headers=HEADERS, timeout=15)
    resp.raise_for_status()
    if not resp.encoding or resp.encoding.lower() == "iso-8859-1":
        resp.encoding = resp.apparent_encoding or "utf-8"
    return BeautifulSoup(resp.text, "html.parser")


# 第 1 题：读取页面并观察节点结构 
print(f"实际使用 URL: {URL}")
print(f"访问日期:     {datetime.date.today().isoformat()}\n")

soup = _fetch_soup(URL)
weather_node = soup.select_one('div[id="7d"] ul.t.clearfix')

if weather_node is not None:
    all_children = list(weather_node.children)
    tag_children = [c for c in all_children if isinstance(c, Tag)]

    print("【第 1 题：节点结构观察】")
    print("定位到的节点:", f"<{weather_node.name} class={weather_node.get('class')}>")
    print("直接子节点总数（含空白文本节点）:", len(all_children))
    print("直接 Tag 子节点数量:", len(tag_children))
    print("直接 Tag 子节点标签名:", [c.name for c in tag_children])
    if tag_children:
        first_li = tag_children[0]
        print("第 1 个 <li> 的直接 Tag 子节点:", [c.name for c in first_li.children if isinstance(c, Tag)])
        print("第 1 个 <li> 内容预览:", _normalize_space(first_li.get_text(" ")))
else:
    print("未找到天气数据节点。")


#  第 2 题：retrieve_climate_table(s) 
def retrieve_climate_table(s: str) -> Tag | None:
    """返回衡阳页面中的天气数据 Tag 对象；找不到返回 None；请求出错抛异常。"""
    page_soup = _fetch_soup(s)
    node = page_soup.select_one('div[id="7d"] ul.t.clearfix')
    if node is not None and node.find("li"):
        return node
    return None


# 第 3 题：list_climate_table_row_names(s) =
def list_climate_table_row_names(s: str) -> list[str] | None:
    """返回去除首尾空白的纯 Python 字符串行名列表；找不到返回 None。"""
    node = retrieve_climate_table(s)
    if node is None:
        return None

    row_names: list[str] = []
    for li in node.find_all("li", recursive=False):
        h1 = li.find("h1")
        if h1 is not None:
            name = _normalize_space(h1.get_text(" ", strip=True))
            if name:
                row_names.append(str(name))
    return row_names


print("\n【第 2 题：retrieve_climate_table】")
tbl = retrieve_climate_table(URL)
print("返回值类型:", type(tbl).__name__ if tbl is not None else None)
if tbl is not None:
    print("返回节点:", f"<{tbl.name} class={tbl.get('class')}>")

print("\n【第 3 题：list_climate_table_row_names】")
rows = list_climate_table_row_names(URL)
print("行名列表:", rows)
if rows is not None:
    print("全部为纯 str:", all(type(r) is str for r in rows))
```
